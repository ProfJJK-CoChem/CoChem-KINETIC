# Audit 139: Barrierless Radical Recombination
This notebook tests the behavior of CoChem-KINETIC when evaluating barrierless radical recombination, such as Methyl radical recombination, where the apparent activation energy evaluates to a negative value (e.g., -2.0 kcal/mol).

According to the audit requirements, KINETIC must intercept the $E_a < 0$ condition, halt standard TST, and apply the Phase Space Theory centrifugal bottleneck approximation yielding $\approx 3 \times 10^{13} \text{ cm}^3 \text{mol}^{-1} \text{s}^{-1}$.

```python
import sys
import os
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '..')))
from kinetic_core.thermo import calculate_eyring_rate

print("Running Document 139 Audit: Barrierless Radical Recombination")

ea_apparent = -2.0
temp = 298.15

# Reaction order 2 to output bimolecular rate constant
rate = calculate_eyring_rate(delta_g=ea_apparent, temp=temp, reaction_order=2)

print(f"Input Apparent Ea: {ea_apparent} kcal/mol")
print(f"Calculated Rate Constant: {rate:.2e} cm^3 mol^-1 s^-1")

expected_rate = 3e13

assert abs(rate - expected_rate) < 1e12, f"FAIL: KINETIC did not apply the PST bottleneck approximation. Expected ~{expected_rate:.1e}, but got {rate:.1e}"
print("PASS: System dynamically applied PST.")
```
